### Buat Struktur Folder 

```python
from pathlib import Path 
config = 'config'
core = 'diamond_web'
requirements = 'requirements'

# Buat file requirements dasar 
Path(f'{requirements}').mkdir(exist_ok=True)
Path(f'{requirements}/base.txt').touch()
Path(f'{requirements}/dev.txt').touch()
Path(f'{requirements}/prod.txt').touch()

# Buat folder diamond_web 
Path(f'{core}/constants').mkdir(exist_ok=True)
Path(f'{core}/fixtures').mkdir(exist_ok=True)
Path(f'{core}/fixtures/default_templates').mkdir(exist_ok=True)
Path(f'{core}/models').mkdir(exist_ok=True)
Path(f'{core}/models/__init__.py').touch()
Path(f'{core}/forms').mkdir(exist_ok=True)
Path(f'{core}/forms/__init__.py').touch()
Path(f'{core}/views').mkdir(exist_ok=True)
Path(f'{core}/views/__init__.py').touch()
Path(f'{core}/views/tiket').mkdir(exist_ok=True)
Path(f'{core}/views/tiket/__init__.py').touch()
Path(f'{core}/management').mkdir(exist_ok=True)
Path(f'{core}/management/__init__.py').touch()
Path(f'{core}/management/commands').mkdir(exist_ok=True)
Path(f'{core}/management/commands/__init__.py').touch()
Path(f'{core}/static').mkdir(exist_ok=True)
Path(f'{core}/static/assets').mkdir(exist_ok=True)
Path(f'{core}/static/docs').mkdir(exist_ok=True)
Path(f'{core}/static/js').mkdir(exist_ok=True)
Path(f'{core}/static/partials').mkdir(exist_ok=True)
Path(f'{core}/templates').mkdir(exist_ok=True)
Path(f'{core}/templates/__init__.py').touch()
Path(f'{core}/templatetags').mkdir(exist_ok=True)
Path(f'{core}/tests').mkdir(exist_ok=True)
Path(f'{core}/tests/__init__.py').touch()
Path(f'{core}/utils').mkdir(exist_ok=True)
Path(f'{core}/utils/__init__.py').touch()

Path('.env.example.dev').touch()
Path('.gitignore').touch()
```
